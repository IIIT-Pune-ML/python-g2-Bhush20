```python
print(binary)
```

### Q2.

```python
import numpy as np

x = np.array([
    [10,100,5],
    [12,110,6],
    [11,105,7],
    [13,500,6],
    [14,108,8],
    [15,115,9],
])
```

1. Calculate the mean and standard deviation. 

```python
mean = x.mean(axis=0)
std = x.std(axis=0)

print("Mean :  ",mean)
print("Std: ",std)
```

2. Identify values satisfying: 
∣x−μ∣>2σ 

```python
outlier=np.abs(x-mean)>2*std
print(outlier)
```

3.Replace such values with the corresponding column median.

```python
median = np.median(x,axis=0)
print("mediam: ",median)
```

4. Do this without modifying non-outlier values.

```python
xClean=np.where(outlier,median,x)
print(xClean)
```

### Q3.

```python

import pandas as pd  
df = pd.DataFrame({ 'Student': ['A','B','C','D','E','F','G','H'],  
'Department': ['CSE','CSE','ECE','ECE','CSE','ECE','CSE','ECE'],  
'Gender': ['F','M','F','M','M','F','F','M'],  
'Marks': [85,72,91,65,78,88,95,70], 
'Attendance': [92,81,96,72,85,90,98,75] }) 
```

1.Average marks for each department. 

```python
avgMarks=df.groupby('Department')['Marks'].mean()
print(avgMarks)
```

2. Average marks separately for each (Department, Gender) group.

```python
avg_Dept_Gen = df.groupby(['Department','Gender'])['Marks'].mean()
print(avg_Dept_Gen)
```

3. Student with the highest marks in each department.

```python
HigestMarks = df.loc[df.groupby('Department')['Marks'].idxmax()]
print(HigestMarks)
```

4. Students whose marks are above their department average.

```python
deptAvg = df.groupby('Department')['Marks'].transform('mean')
aboveAverage= df[df['Marks']>deptAvg]
print(aboveAverage)
```

5. Department having the highest average attendance.

```python
avgAttendance = df.groupby('Department')['Attendance'].mean()
print(avgAttendance)

Higest_attendance_dept = avgAttendance.idxmax()
print("Dept with higest average attendance is:",Higest_attendance_dept)
```

### Q4.


```python
df = pd.DataFrame({ 'ID': [101,102,103,101,104,102,105],  
'Name': ['A','B','C','A','D','B','E'], 
'Age': [20,21,22,20,23,21,24],  
'Score': [80,75,90,85,88,75,92] })  
print(df)
```

1. Identify duplicate ID's

```python
duplicateId = df[df['ID'].duplicated(keep= False)]['ID'].unique()
print(duplicateId)
```

2. For each ID, determine whether its records are identical. 

```python
for id in duplicateId:
    records = df[df['ID']==id]
    identical= records.drop(columns=['ID']).drop_duplicates().shape==1

    print(f"ID{id}:{'Identical' if identical else 'not identical'}")
```

3. If multiple records exist for an ID, retain the record having the highest 
Score.

```python
clean_df = df.sort_values('Score', ascending=False).drop_duplicates(
    subset='ID',
    keep='first'
)
clean_df = clean_df.sort_values('ID').reset_index(drop=True)
```

4. Produce a clean DataFrame containing exactly one record per ID.

```python
print("\nClean DataFrame:")
print(clean_df)
```

5. Report how many records were removed. 

```python
records_removed = len(df) - len(clean_df)
print("\nRecords removed:", records_removed)
```

### Q5.

```python
import pandas as pd
df = pd.DataFrame({
    'date' : [
        '2026-01-01',
        '2026-01-02',
        '2026-01-04',
        '2026-01-05',
        '2026-01-08',
        '2026-01-09'
    ],
    'sales' : [100,120,150,130,180,200]
})
```

1. Convert Date into a proper datetime column.

```python
df['date'] = pd.to_datetime(df['date'])
df = df.set_index('date')
```

2. Detect the missing dates. 

```python
full_date = pd.date_range(
    start=df.index.min(),
    end=df.index.max(),
    freq='D'
)
missing_date = pd.Index(full_date).difference(pd.Index(df.index))
print("Missing dates :")
print(missing_date)
```

3. Reindex the DataFrame to contain every date between the minimum and 
maximum date.

```python
df = df.reindex(full_date)
```

4. Keep missing sales as NaN.

```python
print("dataFrame after reindexing:")
print(df)
```

5. Calculate a 3-day rolling average of sales. 

```python
df['Rolling_Avg'] = df['sales'].rolling(
    window=3,
    min_periods=1
).mean()

print("\nDataFrame with rolling average:")
print(df)
```

6. Identify the date on which the rolling average is maximum.

```python
max_date = df['Rolling_Avg'].idxmax()
max_value = df['Rolling_Avg'].max()

print("\nMaximum rolling average:")
print("Date:", max_date)
print("Rolling Average:", max_value)
```

### Q6.

```python
import numpy as np

P = np.array([
    [1, 2],
    [4, 6],
    [7, 3],
    [2, 8],
    [9, 5]
])

diff = P[:, np.newaxis, :] - P[np.newaxis, :, :]

D = np.sqrt(np.sum(diff ** 2, axis=2))

print("Distance Matrix:")
print(D)

i, j = np.unravel_index(np.argmax(D), D.shape)

print("\nMaximum distance:", D[i, j])
print("Pair of points:", i + 1, "and", j + 1)
print("Points:", P[i], "and", P[j])
```
